```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

## 1. Load the dataset

The file is named `.xls` but it is actually CSV text, so we use `pd.read_csv`.

```python
dataset = pd.read_csv("ObservationData_lavlqce.xls")

print("Shape (rows, columns):", dataset.shape)
dataset.head()
```

## 2. Inspect columns and countries

```python
list(dataset.columns)
```

```python
dataset["Country"].unique()
```

```python
dataset.info()
```

## 3. Filter to Nigeria

Keep a smaller set of useful columns, then isolate Nigeria.

```python
selected_columns = [
    "Country",
    "Year",
    "Real GDP growth (annual %)",
    "Inflation, consumer prices (annual %)",
    "Exports of goods and services (% of GDP)",
    "Imports of goods and services (% of GDP)",
    "Gross capital formation (% of GDP)",
    "Current account balance (Net, BoP, cur. US$)",
    "General government final consumption expenditure (% of GDP)",
    "Household final consumption expenditure  (% of GDP)",
]

dataset_selected = dataset[selected_columns]
nigeria = dataset_selected[dataset_selected["Country"] == "Nigeria"].copy()

print("Nigeria rows:", nigeria.shape[0])
nigeria.head()
```

## 4. Clean missing values

```python
print("Missing values before:")
print(nigeria.isna().sum())

nigeria = nigeria.dropna()

print("\nMissing values after:")
print(nigeria.isna().sum())
print("\nClean shape:", nigeria.shape)
```

## 5. Quick look at the numbers

```python
nigeria.describe()
```

## 6. Simple plots

GDP growth and inflation over time.

```python
plt.figure(figsize=(10, 4))
plt.plot(nigeria["Year"], nigeria["Real GDP growth (annual %)"], marker="o")
plt.title("Nigeria — Real GDP Growth (annual %)")
plt.xlabel("Year")
plt.ylabel("Real GDP growth (annual %)")
plt.grid(True, alpha=0.3)
plt.show()
```

```python
plt.figure(figsize=(10, 4))
plt.plot(nigeria["Year"], nigeria["Inflation, consumer prices (annual %)"], marker="o", color="tab:orange")
plt.title("Nigeria — Inflation (consumer prices, annual %)")
plt.xlabel("Year")
plt.ylabel("Inflation (annual %)")
plt.grid(True, alpha=0.3)
plt.show()
```

```python
plt.figure(figsize=(6, 5))
plt.scatter(
    nigeria["Inflation, consumer prices (annual %)"],
    nigeria["Real GDP growth (annual %)"],
)
plt.title("Inflation vs Real GDP Growth")
plt.xlabel("Inflation (annual %)")
plt.ylabel("Real GDP growth (annual %)")
plt.grid(True, alpha=0.3)
plt.show()

corr = nigeria["Real GDP growth (annual %)"].corr(
    nigeria["Inflation, consumer prices (annual %)"]
)
print("Correlation (GDP growth vs inflation):", round(corr, 3))
```
